```python
def get_distance(point1,point2):
    pre_dist=0
    for i in range(len(point1)):
        pre_dist= pre_dist+(point2[i]-point1[i])**2
    return pre_dist**0.5
    
    
```

$ Delivery Route Planning $

A food delivery company needs to optimize their routes. They have a restaurant at coordinates (2, 5) and need to deliver to three customers:

Customer A at (8, 13)
Customer B at (10, 1)
Customer C at (-1, 9)

In what order should they make the delivery to reduce time of delivery?

```python
source = (2, 5) 
destinations ={
    "A":(8, 13),
    "B":(10, 1),
    "C":(-1, 9)
}

for destination in destinations.items():
    print(destination)
    break
    distance=get_distance(source,destination[1])
    print(f"Distance to {destination[0]} is {distance}")
```

Based on the distances, we need to deliver to C, then B and finally A.

```python
source = (2, 5) 
destinations =[(8, 13),(10, 1),(-1, 9)]
customers=["A","B","C"]
   
for i,destination in enumerate(destinations):
    distance=get_distance(source,destination)
    print(f"Distance for {customers[i]} is {distance}")
```
