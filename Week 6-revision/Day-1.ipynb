```python
prices =[4.95,9.95,14.95,19.95,24.95]
print("Orginal Price discounted amount New Price")
for price in prices:
    discount_amount= round(0.6*price,2)
    new_price = round(price-discount_amount,2)
    print(f"${price} ${discount_amount} ${new_price}")
    
```

```python
prices =[4.95,9.95,14.95,19.95,24.95]
print("Orginal Price\t | discounted amount\t | New Price")
for price in prices:
    discount_amount= round(0.6*price,2)
    new_price = round(price-discount_amount,2)
    print(f"${price}\t\t | ${discount_amount}\t\t| ${new_price}")
    
```

```python
prices =[4.95,9.95,14.95,19.95,24.95]
print("Orginal Price\t| discounted amount\t| New Price")
for price in prices:
    discount_amount= round(0.6*price,2)
    new_price = round(price-discount_amount,2)
    print(f"${price:15}| ${str(discount_amount):21}| ${str(new_price)}")
    
```
