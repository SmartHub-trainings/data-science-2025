```python
from random import randint
class Bank_Account:
    def __init__(self,acc_name,dob,state,email,country,phone,init_deposit):
        print("This is being called")
        self.acc_name=acc_name
        self.dob=dob
        self.email=email
        self.country=country
        self.phone=phone
        self.init_deposit=init_deposit
        self.state=state
        self.acc_balance=init_deposit
        self.acc_type="Savings"
        self.acc_number="144"
        for i in range(7):
            digit= randint(0,9)
            self.acc_number = self.acc_number+str(digit)
    
     
    def make_withdrawal(self,amount):
        if amount <self.acc_balance:
            self.acc_balance =self.acc_balance-amount
        else:
            print("Insufficient funds")
        
        
            
            
```
