```python
list_of_numbers =input("Enter numbers, separated by comma: ")
print(list_of_numbers)
list_splitted=list_of_numbers.split(",")
numbers =[]
for number in list_splitted:
    numbers.append(float(number))
menu ="""What operation do you want to perform?
1. Summation
2. Average
3. Mode
4. Median\n"""
while True:
    response=input(menu)
    if response not in "1234":
        print("Invalid option")
    elif response=="1":
        result=my_sum(numbers)
        print(f"The sum is {result}")
    elif response=="2":
        result=average(numbers)
        print(f"The average is {result}")
    continuation_response = input("Do you wish to perform another operation on this set?(y/n)").lower()
    if continuation_response in ("n","nah","no","nope"):
        break
    
    
```

```python
def median(numbers):
    length =len(numbers)
    numbers.sort()
    if length%2==0:
        pass
    else:
        pos= length//2
        return numbers[pos]
```

```python
median([2,10,20,-1,5])
```

```python
def median(numbers):
    length =len(numbers)
    numbers.sort()
    if length%2==0:
        target_pos=length//2
        return (numbers[target_pos-1]+numbers[target_pos])/2
    else:
        pos= length//2
        return numbers[pos]
```

```python
median([25,2,10,30])
```

```python
def median(numbers):
    length =len(numbers)
    my_numbers =list(numbers)
    my_numbers.sort()
    if length%2==0:
        target_pos=length//2
        return (my_numbers[target_pos-1]+my_numbers[target_pos])/2
    else:
        pos= length//2
        return my_numbers[pos]
```
