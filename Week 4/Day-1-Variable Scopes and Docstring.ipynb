```python
##  LEGB rule
Local Scope
Enclosed Function
Global Scope
Built-in function
```

```python
y=20
def outer_func(a):
    y=10
    def inner_func(b):
        y=1
        z= a +b+y
        print(z)
    inner_func(5)
    print(y)

print(y)
```

```python
outer_func(8)
```
