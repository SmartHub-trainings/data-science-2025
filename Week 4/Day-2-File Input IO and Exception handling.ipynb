```python
from random import randint
names=["Smith","Esther","Clayton","Charles","Okon","ChemChris"]
scores_file=open("scores.txt","w")
scores_file.write("Name,score\n")
for name in names:
    score = randint(0,100)
    scores_file.write(f"{name},{score}\n")
scores_file.close()
    
```
