## File input and output

```python
open("names.txt","w")
```

```python
file_handler=open("names.txt","w")
file_handler.write("Smith")
file_handler.close()
```

```python
file_handler=open("names.txt","w")
file_handler.write("Clayton")
file_handler.close()
```

```python
file_handler=open("names.txt","a")
file_handler.write("Okon the traveller")
file_handler.close()
```

```python
file_handler=open("names.txt","a")
file_handler.write("\nCharles")
file_handler.close()
```

```python
file_handler=open("names.txt","r")
file_handler.read()
```

```python
file_handler=open("names.txt","r")
print(file_handler.readlines())
```

```python
file_handler=open("names.txt","r")
print(file_handler.readline())
```
