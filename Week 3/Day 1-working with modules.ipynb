```python
magic_number = random.randint(1,10)
for i in range(6):
    user_guess = int(input("Give a shot!: "))
    if user_guess ==magic_number:
        print("You did good")
        break
    else:
        # this is a nested if statement.
        # i.e an if with another if
        if i !=5:
            print("Wrong guess\nTry again")
        else:
            print("game Over!!\nGet Some more cash from mum's bag")
```
