```python
names ={"Chemchris","Charles","Okon","Esther","Smith"}

```

```python
def count_distict(name):
  letters =[]
  for letter in name.lower():
    if letter not in letters:
      letters.append(letter)
  return len(letters)


```

```python
count_distict("Smith")
```

```python
count_distict("Chemchris")
```

```python
def count_distict(name):
  return len(set(name.lower()))



```

```python
len(set("Chemchris".lower()))
```

```python
relation=set()
for name in names:
  relation.add((name,count_distict(name)))
print(relation)
```

```python
def func(x):
  return 2*x**2+5
```

```python
func(5)
```

```python
def get_range(numbers):
  range =set()
  for number in numbers:
    range.add(func(number))
  return range
```

```python
get_range({1,5,7,9})
```

```python
def get_pairs(numbers):
  pairs =set()
  for number in numbers:
    pairs.add((number,func(number)))
  return pairs
```

```python
get_pairs({1,5,7,9})
```
