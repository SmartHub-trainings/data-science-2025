## Groupby and Pivot table

```python
sales_data = pd.read_csv("../datasets/store/sales_data.csv")
```

```python
sales_data
```

```python
sales_data.sales.sum()
```

```python
sales_data.sales.mean()
```

```python
sales_data.store.unique()
```

```python
sales_data.groupby("store")
```

```python
sales_data.groupby("store").sum()
```

```python
sales_data.groupby("store").count()
```

```python
sales_data.groupby("store").mean()
```

```python
sales_data.groupby("store").max()
```

```python
sales_data["product"].sort_values()
```

```python
sales_data.groupby("store")["sales"].max()
```

```python
sales_data.groupby("store")[["sales","units"]].mean()
```

```python
sales_data.groupby("store")[["sales","units"]].mean().sort_values("sales",ascending=False)
```

```python
sales_data\
.groupby("store")[["sales","units"]]\
.mean()\
.plot(kind="barh");
```

```python
sales_data.groupby("product")[["sales"]].mean()
```

```python
sales_data.groupby(["store","product"])[["sales"]].mean()
```

```python
sales_data.groupby(["store","product"])[["sales"]].mean().unstack()
```
