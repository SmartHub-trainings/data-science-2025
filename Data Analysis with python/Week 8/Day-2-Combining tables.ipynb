```python
import pandas as pd
import numpy as np
```

```python
table1=pd.DataFrame({"name":["Charles","Clayton","Esther","Okon","ChemChris"],
             "age":np.random.randint(20,60,5)})
```

```python
table2=pd.DataFrame({"name":["Clayton","Esther","ChemChris"],
             "state":["Delta","Imo","Delta"]})
```

```python
table1
```

```python
table2
```

```python
pd.merge(table1,table2)
```

```python
pd.merge(table1,table2,how="inner")
```

```python
pd.merge(table1,table2,how="inner", on="age")
```

```python
table1["isMarried"]=pd.Series([True,True,False,False,True])
table2["isMarried"]=pd.Series([True,True,True])
```

```python
table1
```

```python
table2
```

```python
pd.merge(table1,table2)
```

```python
pd.merge(table1,table2,on="name")
```

```python
pd.merge(table1,table2,on="isMarried")
```

```python
table1
```

```python
pd.merge(table1,table2,how="left",on="name")
```

```python
pd.merge(table2,table1,how="left",on="name")
```

```python
pd.merge(table1,table2,how="right",on="name")
```

```python
pd.merge(table2,table1,how="outer",on="name")
```
