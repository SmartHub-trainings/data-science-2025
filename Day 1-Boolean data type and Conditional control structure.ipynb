# Control structures

```python
age = int(input("Enter age: "))
if age>=18:
    print("Adult")
```

```python
age = int(input("Enter age: "))
if age>=18:
    print("Adult")
if age<18:
    print("Teen")
```

```python
age = int(input("Enter age: "))
if age>=18:
    print("Adult")
else:
    print("Teen")
```

```python
age = 67
if age<18:
    print("Teenager")
elif age<60:
    print("Adult")
else:
    print("Elderly")
```

```python
score = int(input("Enter a score: "))
if score >=70:
    print("A")
elif score>=60:
    print("B")
elif score>=50:
    print("C")
elif score>=45:
    print("D")
elif score>=40:
    print("E")
else:
    print("F")
```

```python
score = int(input("Enter a score: "))
if score >100:
    print("Invalid Entry")
elif score >=70:
    print("A")
elif score>=60:
    print("B")
elif score>=50:
    print("C")
elif score>=45:
    print("D")
elif score>=40:
    print("E")
elif score>=0:
    print("F")
else:
    print("Invalid Entry")
```
